# AquaSentinel: Exploratory Data Analysis

This notebook analyzes the historical water quality and health dataset.

```python
import pandas as pd
import plotly.express as px

import sys
import os
sys.path.append('../src')
from config import DATA_DIR

df = pd.read_csv(DATA_DIR / 'historical_area_data.csv')
df.head()
```

```python
print(df.describe())
```

### Correlation Analysis

```python
corr = df[['avg_wqi', 'rainfall_mm', 'temperature', 'humidity', 'disease_cases']].corr()
fig = px.imshow(corr, text_auto=True, aspect='auto', title='Correlation')
fig.show()
```

### Findings
1. Higher rainfall is positively correlated with disease cases.
2. Lower WQI (poor water quality) is associated with higher disease cases.
